```python
print("Ingresa un numero entero positivo: ")
num = int(input())

if(num == 0):
    resultado = 0
    print(f"Resultado: {resultado}")
elif(num == 1):
    resultado = 1
    print(f"Resultado: {resultado}")
elif(num > 1):
    a = 0
    b = 1
    for i in range(2, num + 1):
        temporal = a + b + 1
        a = b
        b = temporal
    print(f"Resultado: {b}")
else:
    print("Error: Ingresa un numero valido")
```
